# Defect Measurement Analysis in testIR.py Wood Inspection System

This notebook analyzes the defect measurement methodology used in the testIR.py wood inspection system, specifically examining whether defects are measured using **width (x-dimension)** or **height (y-dimension)** of detection bounding boxes.

## Key Questions:
1. How are bounding box coordinates (x1, y1, x2, y2) processed?
2. Which dimension (width or height) is used for defect size calculation?
3. How do camera calibration factors affect measurements?
4. What are the differences between top and bottom camera measurements?

```python
# Import Required Libraries
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
import pandas as pd
import re
from pathlib import Path

# Configure matplotlib for better visualizations
plt.style.use('seaborn-v0_8')
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12

print("✅ Libraries imported successfully")
print("📊 Ready to analyze defect measurement system")
```

```python
# Load and Parse testIR.py Code
testir_path = Path("/home/inspectura/Desktop/InspecturaGUI/testIR/testIR.py")

try:
    with open(testir_path, 'r', encoding='utf-8') as file:
        testir_code = file.read()
    print(f"✅ Successfully loaded testIR.py ({len(testir_code):,} characters)")
    print(f"📁 File path: {testir_path}")
    
    # Count lines
    lines = testir_code.split('\n')
    print(f"📄 Total lines: {len(lines):,}")
    
except FileNotFoundError:
    print(f"❌ Error: Could not find testIR.py at {testir_path}")
    print("Please check the file path and try again.")
    testir_code = ""
```

```python
# Extract Defect Measurement Functions
def extract_function(code, function_name):
    """Extract a function definition from code"""
    pattern = rf'def {function_name}\s*\([^)]*\):[^}]*?(?=\n    def|\nclass|\n\n\w|\Z)'
    match = re.search(pattern, code, re.DOTALL | re.MULTILINE)
    return match.group(0) if match else f"Function '{function_name}' not found"

# Extract both calculate_defect_size functions
print("🔍 Extracting defect measurement functions...")

# Function 1: ColorWoodDetector.calculate_defect_size
colorwood_func = extract_function(testir_code, "calculate_defect_size")
app_func_pattern = r'def calculate_defect_size\(self, detection_box, camera_name="top"\):[^}]*?(?=\n    def|\nclass|\n\n\w|\Z)'
matches = re.findall(app_func_pattern, testir_code, re.DOTALL)

print(f"📊 Found {len(matches)} calculate_defect_size function(s)")

if matches:
    # Display the key parts of the function
    func_code = matches[0]  # First occurrence (ColorWoodDetector)
    print("\n" + "="*80)
    print("🎯 KEY DEFECT MEASUREMENT LOGIC (from calculate_defect_size)")
    print("="*80)
    
    # Extract the critical lines
    critical_lines = []
    for line in func_code.split('\n'):
        if any(keyword in line for keyword in ['width_px', 'height_px', 'defect_size_px', 'x2 - x1', 'y2 - y1']):
            critical_lines.append(line.strip())
    
    for line in critical_lines:
        print(f"📏 {line}")
        
print(f"\n✅ Analysis complete - {len(critical_lines)} critical measurement lines found")
```
